```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

file_path = os.path.join(path, "data.csv")

df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)

display(df.head())
```

```python
l1 =df['price'].quantile(0.25)
l2=df['price'].quantile(0.75)
IQR=l2-l1
print(IQR)
LR=1.5+IQR*l1
HR=1.5-IQR*l2
print(HR-LR)
outliers = df[
        (df['price'] < LR) |
        (df['price'] > HR)
    ]
print(outliers)
```

```python
# Nominal and ordinal variables
categorical_cols = [
    'condition',
    'view',
    'city',
    'waterfront'
]

for col in categorical_cols:

    plt.figure(figsize=(10, 5))

    df[col].value_counts().head(15).plot(
        kind='bar'
    )

    plt.title(f'Bar Plot of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
for col in outliers:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[col])

    plt.title(f'Box Plot Before Outlier Treatment: {col}')
    plt.show()
```

```python
hist_cols = [
    'price',
    'sqft_living',
    'sqft_lot',
    'sqft_above',
    'sqft_basement'
]

for col in hist_cols:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df[col],
        bins=30,
        edgecolor='black'
    )

    plt.title(f'Histogram Before Normalization: {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.show()
```

```python
# Select numeric features
from sklearn.preprocessing import MinMaxScaler, StandardScaler
scaling_cols = [
    'price',
    'bedrooms',
    'bathrooms',
    'sqft_living',
    'sqft_lot',
    'floors',
    'waterfront',
    'view',
    'condition',
    'sqft_above',
    'sqft_basement',
    'yr_built',
    'yr_renovated'
]

# Create copy
df_minmax = df.copy()

# Apply Min-Max normalization
minmax_scaler = MinMaxScaler()

df_minmax[scaling_cols] = minmax_scaler.fit_transform(
    df[scaling_cols]
)

print("Min-Max Normalization Completed")

display(df_minmax.head())
```

```python
# Create copy
df_zscore = df.copy()

# Apply Z-score standardization
zscore_scaler = StandardScaler()

df_zscore[scaling_cols] = zscore_scaler.fit_transform(
    df[scaling_cols]
)

print("Z-score Standardization Completed")

display(df_zscore.head())
```

```python
for col in hist_cols:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df_minmax[col],
        bins=30,
        edgecolor='black'
    )

    plt.title(f'Histogram After Min-Max Normalization: {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.show()
```

```python
for col in hist_cols:

    plt.figure(figsize=(8, 5))

    plt.hist(
        df_zscore[col],
        bins=30,
        edgecolor='black'
    )

    plt.title(f'Histogram After Z-score Standardization: {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['sqft_living'],
    df['price'],
    alpha=0.5
)

plt.title('Before Normalization')
plt.xlabel('Living Area (sqft)')
plt.ylabel('House Price')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df_minmax['sqft_living'],
    df_minmax['price'],
    alpha=0.5
)

plt.title('After Min-Max Normalization')
plt.xlabel('Normalized Living Area')
plt.ylabel('Normalized House Price')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df_zscore['sqft_living'],
    df_zscore['price'],
    alpha=0.5
)

plt.title('After Z-score Standardization')
plt.xlabel('Standardized Living Area')
plt.ylabel('Standardized House Price')

plt.show()
```

```python
comparison = pd.DataFrame({
    'Original Mean': df[scaling_cols].mean(),
    'MinMax Mean': df_minmax[scaling_cols].mean(),
    'ZScore Mean': df_zscore[scaling_cols].mean(),

    'Original Std': df[scaling_cols].std(),
    'MinMax Std': df_minmax[scaling_cols].std(),
    'ZScore Std': df_zscore[scaling_cols].std()
})

display(comparison)
```

```python
df.to_csv("housing_cleaned.csv", index=False)

df_minmax.to_csv("housing_minmax_normalized.csv", index=False)

df_zscore.to_csv("housing_zscore_standardized.csv", index=False)

print("All preprocessed datasets saved successfully!")
```
